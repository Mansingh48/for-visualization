```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("data visualisation ")

name =  np.array(["jai", "raj" , "rone" , "raya", "raghav"])
course = np.array(["btech"]* 5)
sub1 = np.array([96,97,85,97,98])
sub2 = np.array([90,35,46,56,35])
sub3 = np.array([94,46,57,57,46])
sub4 = np.array([90, 5,57,100,100])
sub5 = np.array([90,90,90,90,50])
fees = np.array(["pending" , "complet","complet","complet","complet"])
study_time = np.array([12,10,8,9,5])
print()

data = {
    "name" : name,
    "course" : course,
    "maths" : sub1,
    "hindi" : sub2,
    "english" : sub3,
    "php" :sub4,
    "coding": sub5,
    "payment" : fees,
    "tabel time": study_time
}
df = pd.DataFrame(data)

print(df.head())

print(df.info())


```

```python

def adding():
    name = input("Enter name: ")
    course = input("Enter course: ")
    maths = int(input("Maths: "))
    hindi = int(input("Hindi: "))
    english = int(input("English: "))
    php = int(input("PHP: "))
    coding = int(input("Coding: "))
    payment = input("Payment: ")
    study_time = int(input("Study time: "))

    df.loc[len(df)] = {
        "name": name,
        "course": course,
        "maths": maths,
        "hindi": hindi,
        "english": english,
        "php": php,
        "coding": coding,
        "payment": payment,
        "tabel time": study_time
    }

    print("Student added successfully!")
adding()
```

```python
# persentage 
# detain 
# headlist 
# attendance

df["persentage"] = ((df["maths"] + df["english"] + df["php"] + df["coding"] + df["hindi"])*100)/500

df["detain list"] = np.where(
    df["payment"] == "complet",
    "allowed",
    "detain",
   
    )

```

```python
def grade(value):
    if value >= 0 and value <= 35:
        return "fail"
    elif value >35 and value <= 60:
        return "b grade"
    elif value > 60 and value <= 85:
        return "A GRADE "
    elif value > 85 and value <= 95:
        return "A+++"
    elif value > 95 and value <= 100:
        return "hacker"
    else:
        return "error"
    
```

```python
df["Grade"] = df["persentage"].apply(grade)
```

```python
counts = df["course"].value_counts()

highest = counts.idxmax()
lowest = counts.idxmin()

ax = sns.countplot(
    x="course",
    data=df
)

for bar in ax.patches:

    if bar.get_height() == counts[highest]:
        bar.set_color("red")

    elif bar.get_height() == counts[lowest]:
        bar.set_color("green")

    else:
        bar.set_color("skyblue")

plt.title("Course Count")
plt.grid(axis="y")

plt.show()
```
